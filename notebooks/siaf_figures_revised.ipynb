# Figure typography — regenerated 16 Nov 2026

Reviewer 2: *"Text on many of the figures is too small and hard to read."*

Every figure is drawn wider than the journal column and scaled down on
inclusion, so what matters is the **rendered** size on the page:

    rendered_pt = size_in_script * (textwidth / figure_width)

with `textwidth = 31pc = 5.147 in` for the sn-jnl single-column layout. A 24-inch
figure at `fontsize=18` therefore lands at **3.9 pt**. An audit of the submitted
set found 77 of 102 measured values below 7 pt and 31 below 5 pt.

Two passes were needed, because `rcParams` governs only text whose size is *not*
passed at the call site:

1. `patch_fonts.py` — scales each script's `rcParams` block to the per-figure
   target.
2. `apply_sweep.py` — raises every **explicit** `fontsize=` / `labelsize=`
   argument, which `rcParams` cannot override. This caught 33 cases across 12 of
   19 scripts, including all ten text elements of `f04.py` and the 3.8 pt family
   labels of `s04.py`.

`font_sweep.py` is the audit itself and can be re-run at any time; it reports
what every explicit size renders at on the page. Backups of each script are kept
as `*.py.prefont_bak` and `*.py.presweep_bak`.

Enlarging text broke layouts that had been tuned for text roughly half the size,
so the following were also changed, each for a stated reason:

| figure | change | reason |
|---|---|---|
| f01 | legend outside; year ticks every 10 yr | legend covered the April/May lines |
| f02 | period labels deferred to a common staggered height, 13 pt | dots are drawn at `xp + 0.66w`, so each label hit the **previous** bar's strip |
| f03 | single `fig.supylabel` | the per-row y-labels collided with each other |
| f04 | titles shortened, colorbar `labelpad` 10 to 42, gutter 0.12 to 0.26 | citations ran together; colorbar labels sat on their tick numbers |
| f06 | one-line tick labels; legend above panel a; x ticks 7.1 pt | two-line labels need ~0.047*fontsize in of slot and had 0.76 |
| f07 | legend 14 pt | 21 entries were overpowering the panel |
| f08 | legend below; subtitles coloured and renamed to SIAF_area / SIAF_drk / SIAF_tot | legend ran through the observed curve |
| s02 | percentages moved below the bars; panel-b legend below | labels no longer fitted inside the segments |
| s04 | in-plot family labels replaced by a legend outside | 18 labels overlapped illegibly, as f07 already solves it |

`f02` keeps the word "Observations" on each period label at 13 pt (6.4 pt on the
page) rather than the 8 pt target: the label text answers a reviewer who read the
submitted "(CMIP6)" / "(Obs-era)" legend as meaning one bar was model output.

```python
# Rendered font size of every figure, measured from the saved PNG rather than
# from figsize (run_one.py saves with bbox_inches='tight', which crops).
from PIL import Image
import glob, os

TEXTWIDTH_IN = 31 * 12 / 72.27          # sn-jnl single column = 5.147 in

for p in sorted(glob.glob('figures/*.png')):
    w, _ = Image.open(p).size
    inches = w / 300.0                  # all figures are saved at 300 dpi
    print(f'{os.path.basename(p):12s} {w:5d}px  {inches:6.2f}in  '
          f'scale {TEXTWIDTH_IN / inches:5.3f}')
```

## Supplementary Figure 3 — Absolute two-sided classification leakage

**Rebuilt 17 Nov 2026. This figure now has an end-to-end generator; it was
previously the only supplied asset in the supplementary set apart from
Supplementary Figure 10.**

Two scripts, kept apart so the figure can be restyled without recomputing:

| script | does |
|---|---|
| `compute_leakage_budget.py` | the decomposition in albedo units -> `leakage_budget.csv` |
| `compute_leakage_flux.py` | the radiative conversion to W m-2 -> `leakage_flux.csv` |
| `s03.py` | draws the figure from `leakage_flux.csv` |

### The decomposition

Epoch means over 1982-1991 and 2015-2024, `PVAL_THRESH = 0.05`,
`SIC_THRESHOLD = 0.15`, with

    dalb_sic = (alpha_ice - alpha_ocn) * dsic      concentration-driven part
    dalb_res = dalb - dalb_sic                     residual part

evaluated twice, once with the month's own ice-ocean contrast and once with the
fixed season-mean contrast 0.54. Albedo follows the pipeline's sensor rule:
CLARA-A3 primary with APP-x gap-fill, **APP-x alone in September**.

### The radiative conversion

`compute_leakage_flux.py` imports `siaf_obs_core.py`, which `make_core.py`
generates as the **verbatim first 588 lines** of `../../python/siaf_obs_analysis.py`
(cut immediately before the ERA5-kernel load and the main loop, with the cut
asserted byte-for-byte). Nothing about the radiative calculation is
reimplemented: RS is `load_RS_ceres_mean` over CERES EBAF 2000-2024,
T07-inverted and scattered to PSN25; the Omega_area / Omega_drk partition and
its NaN structure come from `compute_IS`; the averaging is `domain_mean` with
`AREA_W` over `DOMAIN_PSN`.

The two leakage fields are obtained by passing the decomposition through
`compute_IS` in place of dalpha/dT, so they inherit the published partition
exactly rather than a copy of it:

    blue    compute_IS(dalb_res)[1]  ->  -dalb_res on Omega_area, 0 on Omega_drk
    orange  compute_IS(dalb_sic)[2]  ->  -dalb_sic on Omega_drk,  0 on Omega_area

Each month's point is

    domain_mean(field * RS) * AREA_RATIO / 12

which is the paper's own monthly diagnostic
(`domain_mean(IS * RS) * AREA_RATIO * SEASON_FRAC`) divided by six — the
caption's wording — so the six points sum to a number directly comparable with
SIAF_area and SIAF_drk. These are epoch differences rather than per-kelvin
sensitivities, hence W m-2 and not W m-2 K-1.

### Validation

The rebuild reproduces all six published annual sums to within 0.00006 W m-2,
with the two concentration totals exact:

| | residual | concentration | net |
|---|---|---|---|
| a, monthly contrast — rebuilt | 0.03175 | 0.00701 | 0.02474 |
| a, monthly contrast — published | 0.03181 | 0.00701 | 0.02479 |
| b, fixed 0.54 — rebuilt | 0.00923 | 0.00842 | 0.00081 |
| b, fixed 0.54 — published | 0.00927 | 0.00842 | 0.00085 |

The RS domain means it recovers — 142, 196, 215, 175, 110, 52 W m-2 for
April-September — peak in June with the insolation, which is an independent
check that the CERES path is being read correctly.

**Where the last digit comes from.** The one deliberate departure from
`compute_leakage_budget.py` is September. `compute_IS`, `ARCTIC_SIAL` and
`regrid_sial_to_psn` all require the SIAL grid, so APP-x is mapped onto it with
the pipeline's own nearest-neighbour index `IDX_APPX_TO_SIAL` rather than
zoomed. That is what `siaf_obs_analysis.py` does for September, so the figure
is now consistent with the main results; it moves September's residual from
-0.07923 to -0.07210 in albedo units and accounts for the whole of the
0.00006 W m-2 difference above. April-August are unaffected: the run prints
both sets side by side.

The superseded relative-share version, which hard-coded its numbers, is kept as
`s03_relative_superseded.py`. The two must not be interchanged.

Delivered as `Figures/fs03.png`.

## Supplementary Figures 8-10 — Classification masks, split by month

Reviewer 2 asked that the 36-panel classification plot be broken up: *"Consider
splitting Extended Data Figure 8 as it is very hard to read 36 small panels."*
It now is. `s07.py`, `s08.py` and `s09.py` — the three-way split that had been
written earlier and never adopted — are the delivered Supplementary Figs. 8, 9
and 10, two melt-season months each against all three p-value thresholds, with
the Observational Era block above and CMIP6 Historical below.

**Why splitting, and not a bigger canvas or a rotation.** For a grid figure
printed at column width the on-page type size is capped at about

    741 / (columns x title characters)

which does not depend on the drawn canvas size, because the largest font a
panel can hold scales with the canvas in exactly the way the page scaling
undoes. Rotating does not help either when the figure is portrait: for the
unified masks the binding term was the page height both upright and sideways.
Fewer panels per figure is the only lever that moves it. The unified figure
printed at about **5 pt**; each split figure prints at **8.2 pt**.

The unified generator is kept as `s07_unified.py` for reference, but it is no
longer the delivered figure.
